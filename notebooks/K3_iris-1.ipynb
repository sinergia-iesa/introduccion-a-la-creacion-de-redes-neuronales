# Tu primera red neuronal: clasificar flores de iris

**Sección 3.8 del libro**

Una red aprende a reconocer la especie de una flor (*Setosa*, *Versicolor* o *Virginica*) a partir de 4 medidas de sus hojas. No le damos reglas: aprende de ejemplos.

> **Basado en:** Karatas, M. (2024). *Developing AI Applications: An Introduction*, cap. 3 «Artificial Neural Networks». Rheinwerk Publishing. Código original del autor, adaptado a Google Colab y comentado en español con fines educativos.


## 1. Importar las librerías

```python
import tensorflow as tf
import pandas as pd
from sklearn.model_selection import train_test_split
```

## Cargar los datos

Antes de ejecutar esta celda, sube **iris.csv** al panel de archivos de Colab (ícono de carpeta → botón de subir). Si no lo tienes a mano, usa la línea comentada: lee el archivo directamente desde la página de la presentación.

```python
path = "iris.csv"
# path = "https://sinergia-iesa.github.io/introduccion-a-la-creacion-de-redes-neuronales/data/iris.csv"   # alternativa sin subir el archivo

data = pd.read_csv(path, delimiter=',')
print(data.head())
print("Columnas con huecos:", data.columns[data.isnull().any()])
```

## 2. Preparar los datos

La red solo entiende números, así que la especie (texto) se convierte en 0, 1 y 2.

```python
col_name = 'species'                               # ¿qué queremos predecir?
data[col_name] = data[col_name].astype('category')
data[col_name] = data[col_name].cat.codes          # Setosa→0, Versicolor→1, Virginica→2
print(data[col_name].unique())
```

## 3. Dividir en 4 tablas

Primero se separan **entradas** (`data`) y **respuesta** (`col`). Luego, 80 % para entrenar y 20 % para probar.

| Tabla | Filas | Columnas |
|---|---|---|
| train_data | 120 | 4 |
| train_col | 120 | 1 |
| test_data | 30 | 4 |
| test_col | 30 | 1 |

```python
col = data[col_name]
data = data.drop([col_name], axis=1)

train_data, test_data, train_col, test_col = train_test_split(data, col, test_size=0.2)
print(train_data.shape, train_col.shape, test_data.shape, test_col.shape)
```

## 4. Construir la red

4 entradas → 32 neuronas → 64 neuronas → 3 salidas (una probabilidad por especie).

*En el libro se escribe `input_dim=4` dentro de la primera capa; `tf.keras.Input(shape=(4,))` es la forma actual equivalente y evita advertencias en Colab.*

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(4,)),
    tf.keras.layers.Dense(32, activation='sigmoid'),
    tf.keras.layers.Dense(64, activation='sigmoid'),
    tf.keras.layers.Dense(3, activation='softmax')   # las 3 salidas suman 1
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```

## 5. Entrenar (30 épocas)

Observa cómo `loss` baja y `accuracy` sube.

```python
history = model.fit(train_data, train_col, epochs=30)
```

## 6. Probar con datos que la red nunca vio

```python
test_loss, test_acc = model.evaluate(test_data, test_col)
print('Test accuracy:', test_acc)
```
