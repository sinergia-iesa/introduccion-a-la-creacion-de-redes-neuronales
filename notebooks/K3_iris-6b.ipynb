```python
import tensorflow as tf
import pandas as pd
import numpy as np
import joblib
```

```python
path = "iris_test.csv"
# path = "https://sinergia-iesa.github.io/introduccion-a-la-creacion-de-redes-neuronales/data/iris_test.csv"
data = pd.read_csv(path, delimiter=',')
print(data)
```

Se usa **el mismo** escalador del entrenamiento: `transform`, no `fit_transform`.

```python
s_scaler = joblib.load('scaler1.joblib')
data = s_scaler.transform(data)
le = joblib.load('le1.joblib')

model = tf.keras.models.load_model('model1.keras')
pred = model.predict(data)
print('Probabilidades:', pred)
```

```python
num = np.argmax(pred)                  # posición de la probabilidad más alta
print('Número:', num)
spec = le.inverse_transform([num])    # de número a nombre
print('Especie:', spec)
```
