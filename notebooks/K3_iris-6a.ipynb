```python
import tensorflow as tf
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
import joblib
```

## Cargar los datos

Antes de ejecutar esta celda, sube **iris.csv** al panel de archivos de Colab (ícono de carpeta → botón de subir). Si no lo tienes a mano, usa la línea comentada: lee el archivo directamente desde la página de la presentación.

```python
path = "iris.csv"
# path = "https://sinergia-iesa.github.io/introduccion-a-la-creacion-de-redes-neuronales/data/iris.csv"   # alternativa sin subir el archivo

data = pd.read_csv(path, delimiter=',')
print(data.head())
print("Columnas con huecos:", data.columns[data.isnull().any()])
```

```python
le = LabelEncoder()
col = le.fit_transform(data['species'])   # recuerda qué número es cada especie
data = data.drop(['species'], axis=1)

s_scaler = StandardScaler()
data = s_scaler.fit_transform(data)   # recuerda media y desviación

train_data, test_data, train_col, test_col = train_test_split(data, col, test_size=0.2, random_state=42)
```

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(4,)),
    tf.keras.layers.Dense(32, activation='sigmoid'),
    tf.keras.layers.Dense(64, activation='sigmoid'),
    tf.keras.layers.Dense(3, activation='softmax')
])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(train_data, train_col, epochs=100, verbose=0)
print('Test accuracy:', model.evaluate(test_data, test_col, verbose=0)[1])
```

## Guardar

*El libro usa `model1.h5`; en las versiones actuales de Keras se recomienda `.keras`.*

```python
model.save('model1.keras')
joblib.dump(s_scaler, 'scaler1.joblib')
joblib.dump(le, 'le1.joblib')
print('Listo: model1.keras, scaler1.joblib, le1.joblib')
```
